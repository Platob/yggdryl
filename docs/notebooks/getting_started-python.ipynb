## Describe a schema

````python
from yggdryl import DataType, Field

schema = Field(
    "trade",
    DataType.from_fields(
        [
            Field("id", "int64", nullable=False),
            Field("symbol", "utf8"),
            Field("price", DataType.decimal(18, 4), nullable=False),
        ]
    ),
    nullable=False,
)

assert len(schema.data_type) == 3
assert str(schema.data_type[2].data_type) == "decimal128(18,4)"
````

## Attach metadata

````python
from yggdryl import Field

field = Field("symbol", "utf8", metadata={"source": "book"})
field["venue"] = "XPAR"
field.set_parquet_field_id(7)

assert field["source"] == "book"
assert "venue" in field
assert len(field) == 3
assert field.parquet_field_id == 7
````
